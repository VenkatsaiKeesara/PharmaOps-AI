```python
import pandas as pd
import numpy as np

from pathlib import Path
import warnings
import os

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns",None)
pd.set_option("display.max_rows",100)
pd.set_option("display.float_format","{:,.2f}".format)

print("="*80)
print("ALL REQUIRED LIBRARIES LOADED SUCCESSFULLY")
print("="*80)
```

```python
ai_folder = Path("../../outputs/ai_reports")

ai_folder.mkdir(

    parents=True,

    exist_ok=True

)

print("="*80)
print("AI REPORTS OUTPUT FOLDER READY")
print("="*80)

print(ai_folder.resolve())
```

```python
print("="*80)
print("LOADING FEATURE STORE")
print("="*80)

feature_store = pd.read_csv(

    "../../outputs/processed_data/Feature_Store.csv"

)

print(f"Rows    : {feature_store.shape[0]:,}")
print(f"Columns : {feature_store.shape[1]}")

display(feature_store.head())
```

```python
print("="*80)
print("FEATURE STORE VALIDATION")
print("="*80)

validation = pd.DataFrame({

    "Metric":[

        "Rows",

        "Columns",

        "Duplicate Rows",

        "Missing Values"

    ],

    "Value":[

        len(feature_store),

        feature_store.shape[1],

        feature_store.duplicated().sum(),

        feature_store.isnull().sum().sum()

    ]

})

display(validation)

print("="*80)
print("VALIDATION COMPLETED SUCCESSFULLY")
print("="*80)
```

# SECTION A — Inventory Recommendation Engine

## Business Objective

Generate inventory optimization recommendations using business rules.

### Recommendation Categories

- Immediate Restock
- Overstock Reduction
- FEFO Priority
- Inventory Monitoring

```python
print("="*80)
print("GENERATING INVENTORY RECOMMENDATIONS")
print("="*80)

inventory_ai = feature_store.copy()

inventory_ai["Inventory_Recommendation"] = np.select(

    [

        inventory_ai["Low_Stock_Flag"]==1,

        inventory_ai["Near_Expiry_Flag"]==1,

        inventory_ai["Overstock_Flag"]==1,

        inventory_ai["Inventory_Health_Score"]>=80

    ],

    [

        "Immediate Restock Required",

        "Apply FEFO and Prioritize Clearance",

        "Reduce Overstock Through Promotion",

        "Inventory Operating Normally"

    ],

    default="Continue Inventory Monitoring"

)

inventory_ai["Recommendation_Priority"] = np.select(

    [

        inventory_ai["Low_Stock_Flag"]==1,

        inventory_ai["Near_Expiry_Flag"]==1,

        inventory_ai["Overstock_Flag"]==1

    ],

    [

        "Critical",

        "High",

        "Medium"

    ],

    default="Low"

)

display(

inventory_ai[[

"Medicine_ID",

"Generic_Name",

"Quantity_In_Stock",

"Inventory_Health_Score",

"Inventory_Recommendation",

"Recommendation_Priority"

]].head(10)

)
```

```python
inventory_summary=(

inventory_ai

.groupby(

[

"Inventory_Recommendation",

"Recommendation_Priority"

]

)

.size()

.reset_index(name="Medicine_Count")

.sort_values(

"Medicine_Count",

ascending=False

)

)

display(inventory_summary)

print("="*80)
print("INVENTORY RECOMMENDATION ENGINE COMPLETED")
print("="*80)
```

```python
inventory_ai.to_csv(

ai_folder/

"Inventory_Recommendations.csv",

index=False

)

print("="*80)
print("Inventory Recommendations Exported Successfully")
print("="*80)
```

## Business Insights

- Medicines with low stock should be replenished immediately.
- Near-expiry medicines should follow FEFO inventory rotation.
- Overstock medicines require promotional or procurement adjustments.
- Healthy inventory should continue under standard monitoring.

# SECTION B — Sales Recommendation Engine

## Business Objective

Generate sales optimization recommendations using revenue, profitability, and product movement patterns.

### Recommendation Categories

- Marketing Campaign
- Product Bundling
- Pricing Review
- Maintain Strategy

```python
print("="*80)
print("GENERATING SALES RECOMMENDATIONS")
print("="*80)

sales_ai = feature_store.copy()

sales_ai["Sales_Recommendation"] = np.select(

    [

        sales_ai["Revenue_Category"]=="Low",

        sales_ai["Movement_Type"]=="Slow Moving",

        sales_ai["Profit_Margin_Percentage"]<15,

        sales_ai["High_Revenue_Flag"]==1

    ],

    [

        "Increase Marketing Campaign",

        "Bundle With High Demand Medicines",

        "Review Pricing Strategy",

        "Maintain Current Sales Strategy"

    ],

    default="Continue Performance Monitoring"

)

sales_ai["Recommendation_Priority"]=np.select(

[

sales_ai["Revenue_Category"]=="Low",

sales_ai["Movement_Type"]=="Slow Moving",

sales_ai["Profit_Margin_Percentage"]<15

],

[

"High",

"Medium",

"Medium"

],

default="Low"

)

display(

sales_ai[[

"Medicine_ID",

"Generic_Name",

"Total_Revenue",

"Movement_Type",

"Sales_Recommendation",

"Recommendation_Priority"

]].head(10)

)
```

```python
sales_summary=(

sales_ai

.groupby(

[

"Sales_Recommendation",

"Recommendation_Priority"

]

)

.size()

.reset_index(name="Medicine_Count")

.sort_values(

"Medicine_Count",

ascending=False

)

)

display(sales_summary)

print("="*80)
print("SALES RECOMMENDATION ENGINE COMPLETED")
print("="*80)
```

```python
sales_ai.to_csv(

ai_folder/

"Sales_Recommendations.csv",

index=False

)

print("="*80)
print("Sales Recommendations Exported Successfully")
print("="*80)
```

## Business Insights

- Low-revenue medicines should receive targeted marketing support.
- Slow-moving medicines should be bundled with high-demand products.
- Low-profit medicines should undergo pricing review.
- High-revenue medicines should remain procurement priorities.

# SECTION C — Waste Recommendation Engine

## Business Objective

Generate recommendations to reduce medicine waste and financial losses by analyzing waste percentage, waste cost, expiry risk, and inventory behavior.

### Recommendation Categories

- Reduce Procurement
- Improve FEFO Rotation
- Review Inventory Planning
- Continue Monitoring

```python
print("="*80)
print("GENERATING WASTE RECOMMENDATIONS")
print("="*80)

waste_ai = feature_store.copy()

waste_ai["Waste_Recommendation"] = np.select(

    [

        waste_ai["Waste_Percentage"] >= 20,

        waste_ai["Waste_Risk"] == "High",

        waste_ai["Near_Expiry_Flag"] == 1,

        waste_ai["Waste_Cost_Ratio"] >= 10

    ],

    [

        "Reduce Procurement Quantity",

        "Immediate Waste Review Required",

        "Prioritize FEFO Inventory Rotation",

        "Review Inventory Planning"

    ],

    default="Waste Level Acceptable"

)

waste_ai["Recommendation_Priority"] = np.select(

    [

        waste_ai["Waste_Percentage"] >= 20,

        waste_ai["Waste_Risk"] == "High",

        waste_ai["Near_Expiry_Flag"] == 1

    ],

    [

        "Critical",

        "High",

        "Medium"

    ],

    default="Low"

)

display(

    waste_ai[[

        "Medicine_ID",

        "Generic_Name",

        "Waste_Percentage",

        "Waste_Risk",

        "Waste_Recommendation",

        "Recommendation_Priority"

    ]].head(10)

)
```

```python
waste_summary = (

    waste_ai

    .groupby(

        [

            "Waste_Recommendation",

            "Recommendation_Priority"

        ]

    )

    .size()

    .reset_index(name="Medicine_Count")

    .sort_values(

        "Medicine_Count",

        ascending=False

    )

)

display(waste_summary)

print("="*80)
print("WASTE RECOMMENDATION ENGINE COMPLETED")
print("="*80)
```

```python
waste_ai.to_csv(

    ai_folder /

    "Waste_Recommendations.csv",

    index=False

)

print("="*80)
print("Waste Recommendations Exported Successfully")
print("="*80)
```

## Business Insights

- Medicines with high waste percentages should have procurement quantities reduced.
- High waste risk medicines require immediate operational review.
- Near-expiry medicines should follow strict FEFO inventory rotation.
- Waste cost trends should be monitored regularly to reduce financial losses.

# SECTION D — Supplier Recommendation Engine

## Business Objective

Evaluate supplier performance and generate recommendations to improve procurement efficiency and supply chain reliability.

### Recommendation Categories

- Continue Partnership
- Performance Review
- Improve Lead Time
- Supplier Audit

```python
print("="*80)
print("GENERATING SUPPLIER RECOMMENDATIONS")
print("="*80)

supplier_ai = feature_store.copy()

supplier_ai["Supplier_Recommendation"] = np.select(

    [

        supplier_ai["Supplier_Risk"] == "High",

        supplier_ai["Supplier_Performance_Index"] < 70,

        supplier_ai["Supplier_Rating"] < 4,

        supplier_ai["Preferred_Supplier"] == "Yes"

    ],

    [

        "Conduct Supplier Audit",

        "Review Supplier Performance",

        "Improve Supplier Relationship",

        "Continue Strategic Partnership"

    ],

    default="Monitor Supplier Performance"

)

supplier_ai["Recommendation_Priority"] = np.select(

    [

        supplier_ai["Supplier_Risk"] == "High",

        supplier_ai["Supplier_Performance_Index"] < 70,

        supplier_ai["Supplier_Rating"] < 4

    ],

    [

        "Critical",

        "High",

        "Medium"

    ],

    default="Low"

)

display(

    supplier_ai[[

        "Supplier_ID",

        "Supplier_Name",

        "Supplier_Rating",

        "Supplier_Performance_Index",

        "Supplier_Risk",

        "Supplier_Recommendation",

        "Recommendation_Priority"

    ]].head(10)

)
```

```python
supplier_summary = (

    supplier_ai

    .groupby(

        [

            "Supplier_Recommendation",

            "Recommendation_Priority"

        ]

    )

    .size()

    .reset_index(name="Supplier_Count")

    .sort_values(

        "Supplier_Count",

        ascending=False

    )

)

display(supplier_summary)

print("="*80)
print("SUPPLIER RECOMMENDATION ENGINE COMPLETED")
print("="*80)
```

supplier_ai.to_csv(

    "Supplier_Recommendations.csv",

    index=False

)

print("="*80)
print("Supplier Recommendations Exported Successfully")
print("="*80)

## Business Insights

- High-risk suppliers should undergo periodic audits.
- Suppliers with lower performance scores require improvement plans.
- Preferred suppliers should remain key procurement partners.
- Supplier ratings and lead times should be continuously monitored to strengthen supply chain performance.

# SECTION E — Executive Recommendation Engine

## Business Objective

Generate executive-level recommendations by consolidating inventory, sales, waste, and supplier performance indicators into actionable strategic priorities.

These recommendations are intended for senior management and will later be transformed into executive narratives using Gemini AI.

```python
print("="*80)
print("GENERATING EXECUTIVE RECOMMENDATIONS")
print("="*80)

executive_ai = feature_store.copy()

conditions = [

    executive_ai["Executive_Risk_Score"] >= 80,

    executive_ai["Business_Priority"] == "High",

    executive_ai["Inventory_Criticality"] == "Critical",

    executive_ai["Supplier_Risk"] == "High",

    executive_ai["Waste_Risk"] == "High"

]

recommendations = [

    "Immediate Executive Review Required",

    "High Business Priority - Allocate Resources",

    "Critical Inventory Monitoring",

    "Review Supplier Contract",

    "Launch Waste Reduction Initiative"

]

executive_ai["Executive_Recommendation"] = np.select(

    conditions,

    recommendations,

    default="Continue Standard Business Monitoring"

)

executive_ai["Executive_Priority"] = np.select(

    [

        executive_ai["Executive_Risk_Score"] >= 80,

        executive_ai["Business_Priority"] == "High",

        executive_ai["Inventory_Criticality"] == "Critical"

    ],

    [

        "Critical",

        "High",

        "Medium"

    ],

    default="Low"

)

display(

    executive_ai[

        [

            "Medicine_ID",

            "Generic_Name",

            "Executive_Risk_Score",

            "Business_Priority",

            "Executive_Recommendation",

            "Executive_Priority"

        ]

    ].head(10)

)
```

```python
executive_summary = (

    executive_ai

    .groupby(

        [

            "Executive_Recommendation",

            "Executive_Priority"

        ]

    )

    .size()

    .reset_index(name="Count")

    .sort_values(

        "Count",

        ascending=False

    )

)

display(executive_summary)

print("="*80)
print("EXECUTIVE RECOMMENDATION ENGINE COMPLETED")
print("="*80)
```

```python
executive_ai.to_csv(

    ai_folder /

    "Executive_Recommendations.csv",

    index=False

)

print("="*80)
print("Executive Recommendations Exported Successfully")
print("="*80)
```

## Business Insights

- Medicines with high executive risk scores require immediate management attention.
- High-priority products should receive additional operational focus.
- Critical inventory and supplier issues should be addressed proactively.
- Waste reduction initiatives can significantly improve profitability.

# SECTION F — Executive Decision Intelligence Report

## Objective

Prepare a comprehensive executive intelligence report containing business KPIs, operational risks, and actionable insights.

This structured report is designed as the input for Gemini AI to generate a professional executive narrative suitable for senior management.

### Report Contents

- Executive KPIs
- Revenue Overview
- Inventory Overview
- Waste Overview
- Supplier Performance
- High-Risk Medicines
- Waste Hotspots
- Supplier Risk Summary
- Executive Recommendations

```python
print("="*80)
print("BUILDING EXECUTIVE KPI SUMMARY")
print("="*80)

executive_kpis = {

    "Total Medicines": feature_store["Medicine_ID"].nunique(),

    "Total Suppliers": feature_store["Supplier_ID"].nunique(),

    "Total Categories": feature_store["Category_ID"].nunique(),

    "Total Revenue": feature_store["Total_Revenue"].sum(),

    "Inventory Value": feature_store["Inventory_Value"].sum(),

    "Potential Sales": feature_store["Potential_Sales_Value"].sum(),

    "Potential Profit": feature_store["Potential_Gross_Profit"].sum(),

    "Waste Value": feature_store["Total_Waste_Value"].sum(),

    "Average Supplier Rating": round(feature_store["Supplier_Rating"].mean(),2),

    "Average Profit Margin": round(feature_store["Profit_Margin_Percentage"].mean(),2)

}

kpi_df = pd.DataFrame(

    executive_kpis.items(),

    columns=["KPI","Value"]

)

display(kpi_df)
```

```python
print("="*80)
print("TOP 10 HIGH RISK MEDICINES")
print("="*80)

top_risk = (

    feature_store

    .sort_values(

        "Executive_Risk_Score",

        ascending=False

    )

    [[

        "Medicine_ID",

        "Generic_Name",

        "Executive_Risk_Score",

        "Business_Priority",

        "Waste_Risk",

        "Supplier_Risk"

    ]]

    .head(10)

)

display(top_risk)
```

```python
print("="*80)
print("TOP 10 WASTE MEDICINES")
print("="*80)

top_waste = (

    feature_store

    .sort_values(

        "Total_Waste_Value",

        ascending=False

    )

    [[

        "Medicine_ID",

        "Generic_Name",

        "Total_Waste_Value",

        "Waste_Percentage",

        "Waste_Risk"

    ]]

    .head(10)

)

display(top_waste)
```

```python
print("="*80)
print("SUPPLIER RISK SUMMARY")
print("="*80)

supplier_risk_summary = (

    feature_store

    .groupby("Supplier_Risk")

    .agg(

        Medicines=("Medicine_ID","count"),

        Revenue=("Total_Revenue","sum"),

        Waste=("Total_Waste_Value","sum")

    )

    .reset_index()

)

display(supplier_risk_summary)
```

```python
print("="*80)
print("EXECUTIVE RECOMMENDATION SUMMARY")
print("="*80)

recommendation_summary = (

    executive_ai

    .groupby(

        [

            "Executive_Recommendation",

            "Executive_Priority"

        ]

    )

    .size()

    .reset_index(name="Count")

)

display(recommendation_summary)
```

```python
print("="*80)
print("BUILDING GEMINI INPUT")
print("="*80)

prompt = f"""
===========================================================
PHARMAOPS AI - EXECUTIVE DECISION INTELLIGENCE REPORT
===========================================================

You are the Chief Business Intelligence Officer of PharmaOps AI.

Generate a professional executive business report using the information below.

===========================================================
EXECUTIVE KPIs
===========================================================

{kpi_df.to_string(index=False)}

===========================================================
TOP 10 HIGH RISK MEDICINES
===========================================================

{top_risk.to_string(index=False)}

===========================================================
TOP 10 WASTE MEDICINES
===========================================================

{top_waste.to_string(index=False)}

===========================================================
SUPPLIER RISK SUMMARY
===========================================================

{supplier_risk_summary.to_string(index=False)}

===========================================================
EXECUTIVE RECOMMENDATION SUMMARY
===========================================================

{recommendation_summary.to_string(index=False)}

===========================================================
REPORT REQUIREMENTS
===========================================================

1. Executive Summary

2. Key Business Findings

3. Critical Operational Risks

4. Revenue Growth Opportunities

5. Inventory Optimization Strategy

6. Waste Reduction Strategy

7. Supplier Performance Strategy

8. Executive Recommendations

9. Business Action Plan

10. Final Conclusion

The report should be written professionally for senior management.
"""

print(prompt[:3000])
```

```python
with open(

    ai_folder/"Executive_AI_Prompt.txt",

    "w",

    encoding="utf-8"

) as f:

    f.write(prompt)

print("="*80)
print("EXECUTIVE AI PROMPT EXPORTED")
print("="*80)
```
