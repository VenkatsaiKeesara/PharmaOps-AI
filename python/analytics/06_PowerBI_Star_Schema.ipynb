# SECTION D — Date Dimension (Dim_Date)

## Objective

Create a centralized Date Dimension table for Power BI.

This table consolidates all unique dates from:
- Manufacturing Date
- Expiry Date
- Last Restock Date

The Date Dimension enables efficient time-based reporting, trend analysis, and filtering across the Power BI dashboards.

### Primary Key

- Date

### Business Purpose

- Yearly Analysis
- Monthly Analysis
- Quarterly Analysis
- Weekly Analysis
- Time Intelligence

```python
print("="*80)
print("CREATING DATE DIMENSION")
print("="*80)

date_columns = [

    "Manufacturing_Date",

    "Expiry_Date",

    "Last_Restock_Date"

]

all_dates = pd.concat(

    [

        pd.to_datetime(feature_store[col])

        for col in date_columns

    ]

).dropna().drop_duplicates().sort_values()

dim_date = pd.DataFrame({

    "Date": all_dates

})

dim_date["Year"] = dim_date["Date"].dt.year
dim_date["Quarter"] = "Q" + dim_date["Date"].dt.quarter.astype(str)
dim_date["Month"] = dim_date["Date"].dt.month
dim_date["Month_Name"] = dim_date["Date"].dt.month_name()
dim_date["Week"] = dim_date["Date"].dt.isocalendar().week.astype(int)
dim_date["Day"] = dim_date["Date"].dt.day
dim_date["Day_Name"] = dim_date["Date"].dt.day_name()
dim_date["Weekday"] = dim_date["Date"].dt.weekday + 1

print(f"Rows              : {len(dim_date):,}")
print(f"Columns           : {dim_date.shape[1]}")
print(f"Missing Values    : {dim_date.isnull().sum().sum()}")

print("="*80)

display(dim_date.head())
```

```python
print("="*80)
print("DATE DIMENSION VALIDATION")
print("="*80)

duplicates = dim_date["Date"].duplicated().sum()

print(f"Duplicate Dates : {duplicates}")
print(f"Unique Dates    : {dim_date['Date'].nunique():,}")
print(f"Total Records   : {len(dim_date):,}")

if duplicates == 0:
    print("\nSTATUS : PASSED")
else:
    print("\nSTATUS : FAILED")
```

```python
print("="*80)
print("DATE DIMENSION PROFILE")
print("="*80)

dim_date.info()

print()

display(

    dim_date.describe(include="all").T

)
```

```python
dim_date.to_csv(

    dashboard_folder /

    "Dim_Date.csv",

    index=False

)

print("="*80)
print("Dim_Date EXPORTED SUCCESSFULLY")
print("="*80)

print(dashboard_folder / "Dim_Date.csv")
```

# SECTION E — Fact Table (Fact_Medicine_Performance)

## Objective

Create the central Fact Table for the Power BI Star Schema.

Each record represents the performance of one medicine inventory record and contains all measurable business metrics required for analytical reporting.

### Foreign Keys

- Medicine_ID
- Supplier_ID
- Category_ID

### Measures

- Revenue
- Inventory Value
- Profit
- Waste
- Inventory KPIs
- Sales KPIs
- Executive KPIs

This table serves as the primary source for all Power BI visualizations.

```python
print("="*80)
print("CREATING FACT_MEDICINE_PERFORMANCE")
print("="*80)

fact_columns = [

    # Keys
    "Inventory_ID",
    "Medicine_ID",
    "Supplier_ID",
    "Category_ID",

    # Dates
    "Manufacturing_Date",
    "Expiry_Date",
    "Last_Restock_Date",

    # Inventory Measures
    "Quantity_In_Stock",
    "Unit_Cost",
    "Selling_Price",
    "Inventory_Value",
    "Potential_Sales_Value",
    "Potential_Gross_Profit",

    # Sales Measures
    "Total_Revenue",
    "Total_Quantity_Sold",
    "Revenue_per_Transaction",
    "Revenue_per_Unit_Sold",

    # Waste Measures
    "Total_Waste_Value",
    "Total_Quantity_Wasted",
    "Waste_Percentage",
    "Waste_Cost_Ratio",

    # Financial Measures
    "Profit_per_Unit",
    "Profit_Margin_Percentage",
    "Markup_Percentage",
    "Inventory_ROI",
    "Cost_to_Revenue_Ratio",

    # Inventory KPIs
    "Days_to_Expiry",
    "Inventory_Age_Days",
    "Days_Since_Last_Restock",
    "Stock_Difference",
    "Stock_Coverage_Ratio",

    # Business Flags
    "Low_Stock_Flag",
    "Overstock_Flag",
    "Out_of_Stock_Flag",
    "Near_Expiry_Flag",

    # Classifications
    "ABC_Class",
    "Movement_Type",
    "Waste_Risk",
    "Supplier_Risk",
    "Business_Priority",
    "Inventory_Criticality",
    "Stock_Priority",
    "Executive_Priority",

    # Scores
    "Inventory_Health_Score",
    "Sales_Performance_Score",
    "Supplier_Performance_Index",
    "Executive_Risk_Score"

]

fact_medicine_performance = feature_store[fact_columns].copy()

print(f"Rows              : {fact_medicine_performance.shape[0]:,}")
print(f"Columns           : {fact_medicine_performance.shape[1]}")
print(f"Duplicate Rows    : {fact_medicine_performance.duplicated().sum()}")
print(f"Missing Values    : {fact_medicine_performance.isnull().sum().sum()}")

print("="*80)

display(fact_medicine_performance.head())
```

```python
print("="*80)
print("FACT TABLE VALIDATION")
print("="*80)

print(f"Inventory Records : {fact_medicine_performance['Inventory_ID'].nunique():,}")
print(f"Medicines         : {fact_medicine_performance['Medicine_ID'].nunique():,}")
print(f"Suppliers         : {fact_medicine_performance['Supplier_ID'].nunique():,}")
print(f"Categories        : {fact_medicine_performance['Category_ID'].nunique():,}")

print()

print("Primary Key Check")

duplicate_inventory = fact_medicine_performance["Inventory_ID"].duplicated().sum()

print(f"Duplicate Inventory_ID : {duplicate_inventory}")

if duplicate_inventory == 0:
    print("\nSTATUS : PASSED")
else:
    print("\nSTATUS : FAILED")
```

# SECTION F — Executive KPI Table

## Objective

Create a lightweight KPI table for Power BI card visuals.

This table contains high-level business metrics that will be displayed on the Executive Dashboard.

```python
print("="*80)
print("CREATING EXECUTIVE KPI TABLE")
print("="*80)

executive_kpis = pd.DataFrame({

    "KPI":[

        "Total Revenue",
        "Inventory Value",
        "Potential Sales",
        "Potential Profit",
        "Total Waste",
        "Medicines",
        "Suppliers",
        "Categories",
        "Average Profit Margin",
        "Average Supplier Rating"

    ],

    "Value":[

        feature_store["Total_Revenue"].sum(),

        feature_store["Inventory_Value"].sum(),

        feature_store["Potential_Sales_Value"].sum(),

        feature_store["Potential_Gross_Profit"].sum(),

        feature_store["Total_Waste_Value"].sum(),

        feature_store["Medicine_ID"].nunique(),

        feature_store["Supplier_ID"].nunique(),

        feature_store["Category_ID"].nunique(),

        round(feature_store["Profit_Margin_Percentage"].mean(),2),

        round(feature_store["Supplier_Rating"].mean(),2)

    ]

})

display(executive_kpis)
```

```python
executive_kpis.to_csv(

    dashboard_folder /

    "Executive_KPIs.csv",

    index=False

)

print("="*80)
print("Executive_KPIs.csv EXPORTED")
print("="*80)
```

# SECTION G — Dashboard Filters

## Objective

Create reusable lookup tables for Power BI slicers.

These tables improve dashboard usability while avoiding duplicated values.

```python
print("="*80)
print("CREATING DASHBOARD FILTERS")
print("="*80)

dashboard_filters = {

    "Category_Filter": sorted(feature_store["Category_Name"].dropna().unique()),

    "Supplier_Filter": sorted(feature_store["Supplier_Name"].dropna().unique()),

    "ABC_Filter": sorted(feature_store["ABC_Class"].dropna().unique()),

    "Priority_Filter": sorted(feature_store["Business_Priority"].dropna().unique()),

    "Waste_Filter": sorted(feature_store["Waste_Risk"].dropna().unique()),

    "Movement_Filter": sorted(feature_store["Movement_Type"].dropna().unique())

}

dashboard_filters
```

```python
for filter_name, values in dashboard_filters.items():

    pd.DataFrame(

        values,

        columns=[filter_name]

    ).to_csv(

        dashboard_folder /

        f"{filter_name}.csv",

        index=False

    )

print("="*80)
print("ALL DASHBOARD FILTERS EXPORTED")
print("="*80)
```
