# Executive KPI Summary

## Business Objective

Provide senior management with a consolidated overview of the organization's operational and financial performance using key business indicators.

## Business Questions

- How much inventory is currently managed?
- What is the current inventory investment?
- How much revenue has been generated?
- What is the expected sales potential?
- How much financial loss occurred due to waste?
- How many medicines require immediate attention?

```python
executive_kpis = {

    "Total Medicines":
        feature_store["Medicine_ID"].nunique(),

    "Total Suppliers":
        feature_store["Supplier_ID"].nunique(),

    "Inventory Value":
        feature_store["Inventory_Value"].sum(),

    "Potential Sales Value":
        feature_store["Potential_Sales_Value"].sum(),

    "Total Revenue":
        feature_store["Total_Revenue"].sum(),

    "Total Waste Value":
        feature_store["Total_Waste_Value"].sum(),

    "Average Profit Margin (%)":
        feature_store["Profit_Margin_Percentage"].mean(),

    "Average Supplier Rating":
        feature_store["Supplier_Rating"].mean(),

    "Near Expiry Medicines":
        feature_store["Near_Expiry_Flag"].sum(),

    "Low Stock Medicines":
        feature_store["Low_Stock_Flag"].sum()

}

kpi_df = pd.DataFrame(
    executive_kpis.items(),
    columns=["KPI", "Value"]
)

kpi_df
```

```python
plt.figure(figsize=(13,6))

plt.bar(

    kpi_df["KPI"],

    kpi_df["Value"]

)

plt.xticks(rotation=45)

plt.title("Executive KPI Dashboard")

plt.ylabel("Value")

save_chart("01_Executive_KPI_Dashboard")
```

# Financial Analytics

## Business Objective

Evaluate the financial performance of the pharmacy business by analyzing revenue generation, inventory investment, profitability, and return on investment.

## Business Questions

- Which medicine categories generate the highest revenue?
- Which categories contribute the highest inventory investment?
- Which medicines deliver the highest profit?
- Is the current inventory investment generating sufficient returns?

```python
category_revenue = (

    feature_store

    .groupby("Category_Name", as_index=False)

    .agg(
        Total_Revenue=("Total_Revenue","sum")
    )

    .sort_values(
        "Total_Revenue",
        ascending=False
    )

)

category_revenue["Total_Revenue"] = (
    category_revenue["Total_Revenue"]
    .round(2)
)

display(category_revenue)
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    category_revenue["Category_Name"],
    category_revenue["Total_Revenue"]
)

plt.xticks(rotation=45, ha="right")

plt.ylabel("Revenue (₹)")

plt.title("Revenue by Medicine Category")

save_chart("02_Revenue_by_Category")
```

```python
top_profit = (

    feature_store[[
        "Medicine_ID",
        "Generic_Name",
        "Category_Name",
        "Potential_Gross_Profit"
    ]]

    .sort_values(
        "Potential_Gross_Profit",
        ascending=False
    )

    .head(10)

)

top_profit["Potential_Gross_Profit"] = (
    top_profit["Potential_Gross_Profit"]
    .round(2)
)

display(top_profit)
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    top_profit["Generic_Name"],
    top_profit["Potential_Gross_Profit"]
)

plt.xticks(rotation=90)

plt.ylabel("Gross Profit (₹)")

plt.title("Top 10 Most Profitable Medicines")

save_chart("03_Top_Profit_Medicines")
```

```python
inventory_value = (

    feature_store

    .groupby("Category_Name", as_index=False)

    .agg(
        Inventory_Investment=("Inventory_Value", "sum")
    )

    .sort_values(
        by="Inventory_Investment",
        ascending=False
    )

)

# Keep numeric values for charts
inventory_value_chart = inventory_value.copy()

# Format values for display
inventory_value["Inventory_Investment"] = (
    inventory_value["Inventory_Investment"]
    .apply(lambda x: f"₹ {x:,.2f}")
)

print("="*70)
print("INVENTORY INVESTMENT BY CATEGORY")
print("="*70)

display(inventory_value)
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    inventory_value_chart["Category_Name"],
    inventory_value_chart["Inventory_Investment"]
)

plt.xticks(rotation=45, ha="right")

plt.ylabel("Investment (₹)")

plt.title("Inventory Investment by Category")

save_chart("04_Inventory_Investment")
```

```python
financial_summary = pd.DataFrame({

    "Metric":[

        "Total Revenue",

        "Inventory Investment",

        "Potential Sales Value",

        "Potential Gross Profit",

        "Average Profit Margin"

    ],

    "Value":[

        f"₹ {feature_store['Total_Revenue'].sum():,.2f}",

        f"₹ {feature_store['Inventory_Value'].sum():,.2f}",

        f"₹ {feature_store['Potential_Sales_Value'].sum():,.2f}",

        f"₹ {feature_store['Potential_Gross_Profit'].sum():,.2f}",

        f"{feature_store['Profit_Margin_Percentage'].mean():.2f} %"

    ]

})

print("="*65)
print("FINANCIAL KPI SUMMARY")
print("="*65)

display(financial_summary)
```

## Financial Business Insights

### Key Findings

- Revenue generation is concentrated in a limited number of medicine categories, indicating the organization's primary revenue drivers.
- Inventory investment is unevenly distributed across categories, highlighting areas where significant working capital is tied up.
- The profitability analysis identifies medicines with the highest expected gross profit, supporting pricing and promotional strategies.
- Comparing inventory investment with potential sales value helps evaluate whether current stock levels are financially justified.
- Financial KPIs provide management with a consolidated view of revenue generation, inventory utilization, and profit potential.

### Business Value

The financial analysis enables management to identify high-performing product categories, optimize inventory investment, improve pricing decisions, and maximize overall business profitability.

```python
plt.figure(figsize=(12,6))

plt.bar(

    waste_chart["Category_Name"],

    waste_chart["Total_Waste_Value"]

)

plt.xticks(rotation=45, ha="right")

plt.ylabel("Waste Value (₹)")

plt.title("Waste Value by Medicine Category")

save_chart("10_Waste_By_Category")
```

## Business Questions

- Which suppliers perform the best?

```python
supplier_performance = (

    feature_store

    .groupby(

        ["Supplier_ID","Supplier_Name"],

        as_index=False

    )

    .agg(

        Supplier_Rating=("Supplier_Rating","mean"),

        Lead_Time=("Lead_Time_Days","mean"),

        Performance_Index=("Supplier_Performance_Index","mean"),

        Inventory_Value=("Inventory_Value","sum")

    )

    .sort_values(

        by="Performance_Index",

        ascending=False

    )

)

supplier_chart = supplier_performance.copy()

supplier_performance["Inventory_Value"] = (
    supplier_performance["Inventory_Value"]
    .apply(lambda x: f"₹ {x:,.2f}")
)

display(supplier_performance.head(15))
```

```python
supplier_risk = (

    feature_store

    .groupby(

        "Supplier_Risk",

        as_index=False

    )

    .agg(

        Suppliers=("Supplier_ID","nunique")

    )

)

display(supplier_risk)
```

```python
plt.figure(figsize=(12,6))

plt.bar(

    lead_time.head(10)["Supplier_Name"],

    lead_time.head(10)["Average_Lead_Time"]

)

plt.xticks(rotation=45, ha="right")

plt.ylabel("Lead Time (Days)")

plt.title("Top Suppliers by Lead Time")

save_chart("14_Lead_Time")
```

```python
preferred_supplier = (

    feature_store

    .groupby(

        "Preferred_Supplier",

        as_index=False

    )

    .agg(

        Suppliers=("Supplier_ID","nunique"),

        Average_Rating=("Supplier_Rating","mean"),

        Average_Lead_Time=("Lead_Time_Days","mean"),

        Inventory_Value=("Inventory_Value","sum")

    )

)

preferred_display = preferred_supplier.copy()

preferred_display["Inventory_Value"] = (
    preferred_display["Inventory_Value"]
    .apply(lambda x: f"₹ {x:,.2f}")
)

display(preferred_display)
```
