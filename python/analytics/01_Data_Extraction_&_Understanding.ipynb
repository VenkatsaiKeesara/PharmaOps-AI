# Step 5: Data Understanding

In this section, we explore the structure and quality of each dataset by examining:
- Dataset dimensions
- Column names
- Data types
- Missing values
- Duplicate records
- Summary statistics

This helps us understand the data before preprocessing and analysis.

```python
datasets = {
    "Medicines_Master": medicines_df,
    "Medicines_Inventory": inventory_df,
    "Suppliers_Master": suppliers_df,
    "Sales_Transactions": sales_df,
    "Waste_Records": waste_df,
    "Category_Master": category_df
}
```

```python
for name, df in datasets.items():

    print("=" * 70)
    print(f"{name}")
    print("=" * 70)

    print(f"Rows    : {df.shape[0]}")
    print(f"Columns : {df.shape[1]}")

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\n")
```

## Data Types and General Information

This step provides information about each dataset including data types, non-null values, and memory usage.

```python
for name, df in datasets.items():

    print("=" * 70)
    print(f"{name}")
    print("=" * 70)

    df.info()

    print("\n")
```

## Missing Value Analysis

Identify missing values in every dataset to understand data completeness.

```python
for name, df in datasets.items():

    print("=" * 70)
    print(name)
    print("=" * 70)

    missing = df.isnull().sum()

    missing = missing[missing > 0]

    if missing.empty:
        print("No Missing Values\n")
    else:
        print(missing)

    print()
```

## Duplicate Record Analysis

Check for duplicate rows in each dataset.

```python
for name, df in datasets.items():

    duplicates = df.duplicated().sum()

    print(f"{name} : {duplicates} duplicate rows")
```

## Summary Statistics

Generate descriptive statistics for numerical columns to understand distributions and identify potential anomalies.

```python
for name, df in datasets.items():

    print("=" * 70)
    print(name)
    print("=" * 70)

    display(df.describe())
```
